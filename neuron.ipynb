Mistakes I made while building my first neuron
==============================================

1. Using index instead of actual input data

Wrong:
    y = w * i + b

Reason:
    i is just the position (0,1,2,3,4...)
    The neuron should learn from the actual data values.

Correct:
    y = w * x[i] + b


--------------------------------------------------

2. Reinitializing weights every time

Wrong:
    def neuron():
        w = random.uniform(-10,10)

Reason:
    Every time the neuron runs, it gets a brand new weight.
    All learning is lost.

Correct:
    w = random.uniform(-10,10)

    def neuron():
        ...


--------------------------------------------------

3. Prediction list was never cleared

Wrong:
    y1_list = []

    def neuron():
        y1_list.append(...)

Reason:
    Every new prediction gets added to old predictions.

Example:
    First run:
        [2,4,6,8,10]

    Second run:
        [2,4,6,8,10,2.1,4.2,6.3,8.4,10.5]

Correct:
    def neuron():
        y1_list = []


--------------------------------------------------

4. Comparing every y with every prediction

Wrong:

        for j in range(len(pred)):
            error = y[i] - pred[j]

Reason:
    This creates unnecessary comparisons.

Correct:
    error = y[i] - pred[i]

Reason:
    Every prediction should only be compared with its matching target.


--------------------------------------------------

5. Defining a function but never calling it

Wrong:
    def error():
        ...

Reason:
    Python stores the function but does not run it.

Correct:
    error()


--------------------------------------------------

6. Updating weight but never using the new weight

Wrong:
    Predict
    Calculate Error
    Update Weight
    Stop

Reason:
    Learning requires checking whether the new weight improved predictions.

Correct:
    Predict
    Calculate Error
    Update Weight
    Predict Again
    Repeat


--------------------------------------------------

7. No training loop

Wrong:
    Run neuron once

Reason:
    One prediction is not learning.

Correct:
    while True:
        Predict
        Calculate Error
        Update Weight

Reason:
    Learning is repeated correction.


--------------------------------------------------

8. Looking for error == 0

Wrong:
    if error == 0

Reason:
    Decimal values rarely become exactly zero.

Correct:
    if abs(error) < 0.01

Reason:
    Close enough is usually considered learned.


--------------------------------------------------

9. Updating weight for every individual error

Wrong:
    for e in error_list:
        update weight

Reason:
    Positive and negative errors may cancel each other.

Better:
    avg_error = sum(error_list)/len(error_list)

    Update weight using avg_error


--------------------------------------------------

10. Bias was confusing the learning process

Observation:
    My dataset follows:

        y = 2x

Reason:
    The correct bias is already 0.

So during early experiments:
    b = 0

Focus only on learning w.

Later:
    Learn both w and b.


--------------------------------------------------

11. Learning means changing weights

A neuron is not learning if it only predicts.

Prediction:
    y = w*x + b

Learning:
    Calculate Error
    Change w
    Change b
    Predict Again

Without changing weights:
    It is just a calculator.


--------------------------------------------------

12. Basic structure of a learning neuron

Initialize:
    w = random value

Repeat:
    Make Predictions
    Calculate Errors
    Compute Average Error
    Update Weight
    Repeat Until Error Is Small

This is the core idea behind machine learning.

# Corrected version of the same neuron


```python
import random

data = ((1,2), (2,4), (3,6), (4,8), (5,10))

x = [item[0] for item in data]
y = [item[1] for item in data]

# Initialize ONCE
w = random.uniform(-10, 10)
b = 0

def neuron_2():

    global w, b

    while True:

        y1_list = []

        # Prediction
        for i in range(len(x)):
            y1 = w * x[i] + b
            y1_list.append(y1)

        # Error calculation
        error_list = []

        for i in range(len(y)):
            err = y[i] - y1_list[i]
            error_list.append(err)

        avg_error = sum(error_list) / len(error_list)

        print("\n----------------")
        print("w =", round(w, 4))
        print("predictions =", [round(v, 2) for v in y1_list])
        print("errors =", [round(v, 2) for v in error_list])
        print("avg error =", round(avg_error, 4))

        # Stop when learned
        if abs(avg_error) < 0.01:
            print("\nNeuron learned!")
            break

        # Learning step
        if avg_error > 0:
            w += 0.01

        elif avg_error < 0:
            w -= 0.01

neuron_2()
```
